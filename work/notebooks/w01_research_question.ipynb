## 1. My lane (or freestyle) and why

*Name your lane — or say 'freestyle' and describe your own question. One short paragraph: why this one?*

**Lane:** Lane 1 — Content Refresh Prioritization

Content decay is a high-stakes decision for SEO and editorial teams. Updating web pages requires dedicated writer hours. Without predictive prioritization, teams rely on simple heuristics (e.g., flagging any page untouched for 180 days), which yields low precision and wastes team bandwidth on stable pages. Machine learning allows us to combine multi-variable metrics (rank drops, CTR changes, traffic volume) to focus editorial action where it yields the highest return.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Setup path check and load data
import os
import pandas as pd
import numpy as np

# Direct raw URL to  GitHub repo's starter CSV
data_url = "https://raw.githubusercontent.com/salsabielmesl/flyrank-ml-assignments/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(data_url)
print("Data successfully loaded. Dataset shape:", df.shape)
```

## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

* **Unit of Analysis:** An individual page URL (`page_id` level).
* **Decision:** Deciding which decaying web pages should be assigned to the editorial team for a content refresh during the next sprint.
* **Action:** Updating outdated information, re-optimizing keywords, and improving user intent alignment for prioritized pages.
* **Cost of a Wrong Call:**
  * **False Positive (Flagging a healthy page):** Wastes 3–5 hours of editorial time on an unnecessary rewrite, inflating operational overhead.
  * **False Negative (Missing a decaying page):** Allows key organic traffic drivers to continuously drop in search rankings, leading to permanent loss of organic visibility and inbound leads.
* **Why ML Helps:** Traffic decay is non-linear and multi-variable. ML captures interactions across rank positions, historical impressions, and CTR to isolate true traffic loss.

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Label creation
df["is_declining_label"] = df["trend_direction"].str.lower().eq("down").astype(int)
print(f"Total pages: {len(df):,} | Overall declining rate: {df['is_declining_label'].mean():.3f}")
```

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

### Key Dataset Baseline Numbers:
* **Total Pages Analyzed:** 30,000 pages in the anonymized benchmark dataset.
* **Overall Traffic Decay Rate:** 37.6% of pages exhibit organic traffic decline.
* **Heuristic Baseline Precision@50:** A standard rule (pages untouched $\ge$ 180 days with $\ge$ 500 impressions) achieves only **24.0% Precision@50** (38 out of top 50 recommendations are false positives).

```python
# This cell is for CODE (numbers, a query, a check).
# Write your text answer in the cell ABOVE this one — typing sentences here breaks Run All.
# Calculate real numbers from starter data
stale = (df["days_since_last_update"] >= 180).astype(int)
visible = (df["impressions_90d"] >= 500).astype(int)
df["hand_rule_score"] = stale * visible * df["impressions_90d"]

def precision_at_k(scores, labels, k=50):
    order = np.argsort(-np.asarray(scores))
    return np.asarray(labels)[order[:k]].mean()

p50 = precision_at_k(df["hand_rule_score"], df["is_declining_label"], 50)

print(f"1. Dataset Size: {len(df):,} pages")
print(f"2. Baseline Decay Rate: {df['is_declining_label'].mean():.1%}")
print(f"3. Hand-written Rule Precision@50: {p50:.1%}")
```
